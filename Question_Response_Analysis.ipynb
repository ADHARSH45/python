```python
import pandas as pd
df = pd.read_csv("Question_response.csv")
df.dropna(inplace = True)
new_df = df.groupby('Percent').size()
new_df
```

```python
bins = [0,40,50,60,70,80,90,100]
labels = ['<40','<50','<60','<70','<80','<90','<100']
df['group'] = pd.cut(df['Percent'],bins = bins,labels = labels,right = False)
range_count = df['group'].value_counts().sort_index()
range_count
```

```python
import matplotlib.pyplot as plt
plt.xlabel("Number of Students")
plt.ylabel("Range")
plt.bar(labels,range_count,color = ['r','b','y','g'])
plt.show()
```
